# Cross-validation + Hyperparameter Tuning

```python
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_absolute_error

df=pd.read_csv('../data/student_scores.csv')
X=df[['hours_studied','attendance','assignments_completed']]
y=df['score']
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,random_state=42)
```

```python
model=DecisionTreeRegressor(random_state=42)
scores=cross_val_score(model,Xtr,ytr,cv=5,scoring='neg_mean_absolute_error')
print('CV MAE:',-scores.mean())
```

```python
search=GridSearchCV(model,{'max_depth':[1,2,3,4,None]},cv=5,scoring='neg_mean_absolute_error')
search.fit(Xtr,ytr)
print(search.best_params_)
print('test MAE:',mean_absolute_error(yte,search.predict(Xte)))
```
