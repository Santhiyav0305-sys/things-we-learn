# Regression + Classification

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_absolute_error, classification_report

df=pd.read_csv('../data/student_scores.csv')
```

```python
X=df[['hours_studied','attendance']]
y=df['score']
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,random_state=42)
m=LinearRegression().fit(Xtr,ytr)
print('MAE:', mean_absolute_error(yte,m.predict(Xte)))
```

```python
yc=(df['score']>=70).astype(int)
Xtr,Xte,ytr,yte=train_test_split(X,yc,test_size=.25,random_state=42,stratify=yc)
clf=LogisticRegression().fit(Xtr,ytr)
print(classification_report(yte,clf.predict(Xte),zero_division=0) )
```
