# Tiny Neural Network with NumPy
This is educational code designed to show the mechanics of a forward pass and gradient update.

```python
import numpy as np
rng=np.random.default_rng(7)
X=rng.normal(size=(200,2))
y=(X[:,0]*X[:,1]>0).astype(float).reshape(-1,1)
W1=rng.normal(scale=.5,size=(2,4)); b1=np.zeros((1,4))
W2=rng.normal(scale=.5,size=(4,1)); b2=np.zeros((1,1))
```

```python
def sigmoid(z): return 1/(1+np.exp(-z))
def sigmoid_grad(a): return a*(1-a)

for epoch in range(1000):
    h=np.tanh(X@W1+b1)
    p=sigmoid(h@W2+b2)
    loss=-(y*np.log(p+1e-9)+(1-y)*np.log(1-p+1e-9)).mean()
    dp=(p-y)/len(y)
    dW2=h.T@dp; db2=dp.sum(axis=0,keepdims=True)
    dh=dp@W2.T
    dz=dh*(1-h*h)
    dW1=X.T@dz; db1=dz.sum(axis=0,keepdims=True)
    lr=.1
    W1-=lr*dW1; b1-=lr*db1; W2-=lr*dW2; b2-=lr*db2
    if epoch%200==0: print(epoch, loss)
```

```python
pred=(sigmoid(np.tanh(X@W1+b1)@W2+b2)>=.5).astype(int)
print('accuracy:',(pred==y).mean())
```
