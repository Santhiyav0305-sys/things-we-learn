# Pandas + EDA

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/student_scores.csv')
df.head()
```

```python
print(df.info())
print(df.describe())
print(df.isna().sum())
```

```python
plt.scatter(df['hours_studied'], df['score'])
plt.xlabel('Hours studied')
plt.ylabel('Score')
plt.title('Study hours vs score')
plt.show()
```
