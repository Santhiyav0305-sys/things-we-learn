# K-Means + PCA

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

df=pd.read_csv('../data/customers.csv')
X=df[['age','annual_income','spend_score','visits_per_month']]
Xs=StandardScaler().fit_transform(X)
```

```python
labels=KMeans(n_clusters=3,n_init=10,random_state=42).fit_predict(Xs)
print(pd.DataFrame({'cluster':labels}).value_counts().sort_index())
```

```python
Z=PCA(n_components=2).fit_transform(Xs)
plt.scatter(Z[:,0],Z[:,1],c=labels)
plt.xlabel('PC1'); plt.ylabel('PC2'); plt.title('Customer clusters in PCA space'); plt.show()
```
