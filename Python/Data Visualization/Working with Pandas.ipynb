```python
import pandas as pd
# 1.Create student data
students = [
    {"Roll No": 101, "Name": "Arun", "Dept": "CSE", "Mark": 85},
    {"Roll No": 102, "Name": "Bala", "Dept": "ECE", "Mark": 90},
    {"Roll No": 103, "Name": "Chitra", "Dept": "EEE", "Mark": 88},
    {"Roll No": 104, "Name": "Divya", "Dept": "IT", "Mark": 92}
]
# 2.Create DataFrame
df = pd.DataFrame(students)
# Display DataFrame
print(df)
#3. Display first 2 rows
print("\nFirst 2 rows:")
print(df.head(2))
#4. Display last 2 rows
print("\nLast 2 rows:")
print(df.tail(2))
#5. Iterate through rows
print("\nStudent Details:")
for index, row in df.iterrows():
    print("Roll No:", row["Roll No"])
    print("Name:", row["Name"])
    print("Dept:", row["Dept"])
    print("Mark:", row["Mark"])
    print()
# 6.Display data types
print("Data Types:")
print(df.dtypes)
#7. Display statistical summary
print("\nStatistical Summary:")
print(df.describe())
```
